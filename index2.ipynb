```python
#renaming the columns
data = data.rename(columns={'Occupation':'Job'})
data
```

```python
#add new columns 
data.loc[data['Sleep Duration']<=8,'Classification']='Normal'
data.loc[data['Sleep Duration']>8,'Classification']='Adequate'
data
```

```python
data.info()
```

```python
#stripping whitespaces
data.columns = data.columns.str.strip()
data.columns
```

```python
#Visualization 
#Quality of sleep and sleep duration 
#sleep duration and age 
#box plots 
```

```python
#graphical rep
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
x = data['Physical Activity Level']
y = data['Quality of Sleep']
```

```python
plt.scatter(x,y)
plt.title('Quality of sleep vs Physical activity level')
plt.legend
plt.xlabel('Physical activity level')
plt.ylabel('Quality of sleep')
plt.show()
```

```python
#there is a no correlation between quality of sleep and physical activity 
```

```python
x1 = data['Sleep Duration']
y1 = data['Age']
```

```python
plt.bar(x1,y1, edgecolor = 'black', label='age')
plt.title('Age vs Sleep duration')
plt.xlabel('Sleep duration')
plt.ylabel('Age')
plt.legend()
plt.show()
```

```python
x2 = data['Age']
```

```python
plt.hist(x2,bins=10, edgecolor = 'red', label='Sleep duration')
plt.title('Age')
plt.xlabel('number of years')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```
