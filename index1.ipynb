```python
data.loc[data['humidity']<50, 'temperature'] = 'cold'
data.loc[data['humidity']>50, 'temperature'] = 'hot'
data
```

```python
#striping whitespaces from column names
data.columns = data.columns.str.strip()
data.columns
```

```python
#rename columns
data = data.rename(columns ={'humidity':'avg_humidity'})
data
```

```python
data.info()
```

```python
#type to integer
data['meanpressure.int'] = data['meanpressure'].astype('int')
data.info()
```

```python
#converting date column to datetime format
data['date'] = pd.to_datetime(data['date'])
data.info()
```

```python
#using the apply function
def categorize_temp(temp):
    if temp < 20:
        return 'cold'
    elif temp >= 20:
        return 'warm'
    else:
        return 'hot'
```

```python
data['temperature_category'] = data['meantemp'].apply(categorize_temp)
data
```

```python
#.map function
data['month'] = data['date'].dt.month
data
```
